````python
# prompt: perceptron code

import numpy as np

class Perceptron:
    def __init__(self):
        self.weights = None
        self.bias = None

    def set_weights(self, weights):
        self.weights = weights

    def set_bias(self, bias):
        self.bias = bias

    def predict(self, X):
        ##TODO
        ## we call the function (to determine if 0 or 1) with the result of multiplying the inputs (X) times the weights, + bias. np.dot will multiply 
        ## each elements of the array. 
        y_predicted = self._unit_step_func(np.dot(X, self.weights) + self.bias) 
        return y_predicted

    def _unit_step_func(self, x):
        return np.where(x >= 0, 1, 0)

# Example usage
if __name__ == "__main__":
    # Sample data (OR gate)
    X = np.array([[100, -20], [10, 10], [11, 9], [11, 10], [5, 5], [-50,50]])
    y = np.array([1, 1, 1, 1, 0, 0])

    #sample weights and biases
    weights = np.array([0.5, 0.5])
    bias = -10

    # Initialize and train the perceptron
    p = Perceptron()
    p.set_weights(weights)
    p.set_bias(bias)
    # Make predictions
    predictions = p.predict(X)
    print("Predictions:", predictions)
    print("Labels:",y)
````

# Discussion: More on XOR gate

It's not easy to build XOR gate. In fact, it is theoretically impossible to build XOR gate with only one layer of perceptron.

But it is doable with more layers. How?

Think about logic gates, how do you build XOR? (Need a volunteer)


First, let me simplify things a bit by first redefining our Perceptron class to be easier to use

````python
class Perceptron:
    def __init__(self, weights, bias):
        self.weights = np.array(weights)
        self.bias = bias

    def activation(self, x):
        return 1 if x >= 0 else 0

    def predict(self, inputs):
        linear_combination = np.dot(self.weights, inputs) + self.bias
        return self.activation(linear_combination)
````

````python
# Define individual gates using perceptrons
def AND_gate(x1, x2):
    perceptron = Perceptron(weights=[1, 1], bias=-1.5)
    return perceptron.predict([x1, x2])

def OR_gate(x1, x2):
    perceptron = Perceptron(weights=[1, 1], bias=-0.5)
    return perceptron.predict([x1, x2])

def NOT_gate(x):
    perceptron = Perceptron(weights=[-1], bias=0.5)
    return perceptron.predict([x])

##TODO Construct XOR gate using the basic gates
def XOR_gate(x1, x2):
    ##TODO
    # XOR can be constructed using AND, OR, and NOT gates
    # XOR = (x1 AND NOT x2) OR (NOT x1 AND x2)
    not_x1 = NOT_gate(x1)
    not_x2 = NOT_gate(x2)
    and1 = AND_gate(x1, not_x2)
    and2 = AND_gate(not_x1, x2)
    xor_result = OR_gate(and1, and2)
    return xor_result
````

Once you have defined the above function, test it out using the code below.

````python
test_inputs = [(0, 0), (0, 1), (1, 0), (1, 1)]
print("x1 x2 | XOR")
print("-------------")
for x1, x2 in test_inputs:
    result = XOR_gate(x1, x2)
    print(f" {x1}  {x2} |  {result}")
````

# Discussion: Perceptron with learning ability

It is obviously very tedious to calculate and then type out all weights yourself. Wouldn't it be nice if the perceptron can somehow learn the weights and biases from data?

Yep, you don't learn, machine learns. Hence the name, machine learning.

Following is the math formula

image.png

That may not make much sense. Let's do a real example on the board. Using one of the gates from above.

````python
class Perceptron:
    def __init__(self):
        self.weights = None
        self.bias = None
        self.n_iter = 1000
        self.learning_rate = 0.01
    def set_weights(self, weights):
        self.weights = weights

    def set_bias(self, bias):
        self.bias = bias

    def predict(self, X):
        linear_output = np.dot(X, self.weights) + self.bias
        y_predicted = self._unit_step_func(linear_output)
        return y_predicted

    def inspection(self):
        print(f"Weights: {self.weights}")
        print(f"Bias: {self.bias}")

    def _unit_step_func(self, x):
        return np.where(x >= 0, 1, 0)

    def fit(self, X, y):
        """
        Fit the perceptron to the data.

        Parameters:
        X: np.array
            Input features, shape (n_samples, n_features)
        y: np.array
            True labels, shape (n_samples,)
        """
        n_samples, n_features = X.shape
        self.weights = np.zeros(n_features)
        self.bias = 0

        for _ in range(self.n_iter):
            for idx, x_i in enumerate(X):
                # linear_output = np.dot(x_i, self.weights) + self.bias
                # y_predicted = self._unit_step_func(linear_output)
                y_predicted = self.predict(x_i)
                # Update rule
                update = self.learning_rate * (y[idx] - y_predicted)
                self.weights += update * x_i
                self.bias += update
````
