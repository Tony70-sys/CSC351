```python
## This is a function that will take an image as input and and return the number indicating the class of the image. 
import numpy as np

#for i in range(20):
#plt.imshow(x_train[8], cmap='gray')
#    plt.show()
#    print("Label: ", y_train[i])

def predictImage(image): 
    top = np.sum(image[:14, :] > 90) ##we get the sum of the top half of the image, we check if the pixel value is greater than 90, if it is we count it as a dark pixel and we sum all the dark pixels in the top half of the image.
    bottom = np.sum(image[14:, :] > 90) ##we get the sum of the bottom half of the image. 
    left = np.sum(image[:, :14] > 90) ##we get the sum of the left half of the image.
    right = np.sum(image[:, 14:] > 90) ##we get the sum of the right half of the image.
    center = np.sum(image[10:18, 10:18] > 110) ##we get the sum of the center.
    total_dark = np.sum(image > 90) ##we get the sum of all the dark pixels in the image.
    
    ##we start with the "edges" cases, least and most pixels, 8 and 1
    ##if the darkness is less than 72, we return 1, because 1 has the least amount of bright pixels.
    if total_dark < 72:
        return 1

    ##if the darkness is greater than 180, we return 8, because 8 has a big amount of bright pixels.
    if total_dark > 180:
        return 8

    ##after edge cases, we chekc for the top 
    ##if the top has more pixels than the bottom, we return 7
    if top > bottom:
        return 7

    ##if the left has more pixels than the right, we return 6
    if left > right:
        return 6

    ##if the right has more pixels than the left, we return 9
    if right > left:
        return 9

    ##if the center has a good amount of pixels in the center, we return 4.
    if center > 35:
        return 4

    ##if most of the pixels are in the center, it's likely to be a zero
    if center > 45:
        return 0

    ##finally, if the total darkness is less than 105 and skipped the previous if statements, we return 2, because 2 has a low amount of bright pixels.
    if total_dark < 105:
        return 2
    ##and we return 3 if the darness is between 105 and 135 after we check for lower brightness (2).
    if total_dark < 135:
        return 3
    ##otherwise we return 5, which is the only number left.
    else: 
        return 5
    
```

#### DONE:: To get full credit for last task, you will need to test it.

Use a for loop or something, run you function on all images, output the accuracy of your model, which is the percentage of your prediction being correct.

```python
correct = 0

for i in range(len(x_train)):
    pred = predictImage(x_train[i])

    if pred == y_train[i]:
        correct += 1

accuracy = correct / len(x_train)
print(f"Accuracy Percentage: {accuracy * 100:.2f}%")
```
